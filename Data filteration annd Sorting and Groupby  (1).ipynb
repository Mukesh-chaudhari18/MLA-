#### loc and iloc - Indexing 
- loc - Label Based Indexing
- syntax df.loc[row_index,columns_index]

- iloc - Number based indexing
- syntax - df.iloc[row_index,column_index]

```python
user = pd.DataFrame({ 'ID' : [ 1,2,3,4,5,6] , 
                    'Name' : ['Amay','Sakshi','Viraj','Mohini','ravi' ,'vinita'],
                     'salary' : [ 23000,34000,12000,45000,67000,90000],
                     'age' : [ 23,24,26,19,20,21],
                     'job' : [ 'DA' , 'DS', 'MLE', 'DA','MLOPS' ,'DS'] ,
                     'city' : [ 'Pune' , 'Hyd' ,'Ban' ,'Pune' , 'Delhi' ,'Pune']})
```
