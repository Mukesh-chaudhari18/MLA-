# Handling the missing values 

```python
data.isnull().sum()
```

```python
data.columns
```

### To remove missing data - 
- dropna - missing data it drops 

```python
data.dropna(subset=['NAME', 'HOST_IDENTITY_VERIFIED', 'HOST_NAME',
       'NEIGHBOURHOOD_GROUP', 'NEIGHBOURHOOD', 'COUNTRY',
       'COUNTRY_CODE', 'INSTANT_BOOKABLE', 'CANCELLATION_POLICY',
       'CONSTRUCTION_YEAR'] , inplace = True)
```

```python
data.isnull().sum()
```

```python
data['PRICE']
```

```python
data['PRICE'] = data['PRICE'].str.replace('[^0-9]', '', regex=True)
```

```python
data['PRICE']
```

```python
data['PRICE'] = data['PRICE'].astype(float)
```

```python
data['PRICE']
```

#### fillna - function used for filling na values 
data.fillna()

```python
data['PRICE'] = data['PRICE'].fillna(data['PRICE'].mean() )
```

```python
data.isnull().sum()
```

```python
data['SERVICE_FEE']
```

```python
data['SERVICE_FEE'] = data['SERVICE_FEE'].str.replace( '[^0-9]' , '' , regex=True)
```

```python
data['SERVICE_FEE'] = data['SERVICE_FEE'].astype(float)
```

```python
data['SERVICE_FEE'] = data['SERVICE_FEE'].fillna(data['SERVICE_FEE'].mean())
```

```python
data.isnull().sum()
```

```python
data['MINIMUM_NIGHTS'].value_counts()
```

```python
data['MINIMUM_NIGHTS'] =  data['MINIMUM_NIGHTS'].fillna(data['MINIMUM_NIGHTS'].mode().iloc[0])
```

```python
data['MINIMUM_NIGHTS'] = data['MINIMUM_NIGHTS'].astype(int)
```
